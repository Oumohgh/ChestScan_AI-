## 1.2 Charger le dataset

```python
# Etape 1.2 : Telecharger le dataset depuis Kaggle
# Une seule requete : kagglehub telecharge, decompresse et met en cache
# Le meme appel de la partie 2 reutilise ce cache, sans re-telecharger
path = kagglehub.dataset_download("paultimothymooney/chest-xray-pneumonia")
path = Path(path)
```

## 1.3 Explorer la structure des dossiers

```python
# Etape 1.3 : Definir les dossiers de chaque classe
# On ecrit les chemins une fois ici pour ne pas les repeter dans les boucles
# L archive contient elle-meme un sous-dossier chest_xray/


train_dir =  path / "chest_xray" / "chest_xray" / "train"

val_dir =  path / "chest_xray" / "chest_xray" / "val"

test_dir =  path / "chest_xray" / "chest_xray" / "test"


train_classes_paths = [train_dir / "NORMAL", train_dir / "PNEUMONIA"]
val_classes_paths = [val_dir / "NORMAL", val_dir / "PNEUMONIA"]
test_classes_paths = [test_dir / "NORMAL", test_dir / "PNEUMONIA"]


```

## 1.4 Compter les images et leur répartition

```python
# Etape 1.4 : Lister les fichiers image d un dossier
# La fonction renvoie les chemins d une seule classe (NORMAL ou PNEUMONIA)
# On filtre les extensions pour ignorer les fichiers parasites (.DS_Store)
  
allowed_ext = (".jpeg", ".png", ".jpg")
def get_images(images_dir: list, type_index):
    return [
        i for i in images_dir[type_index].rglob("*")
        if i.name.endswith(allowed_ext)
    ]
```

```python
# Etape 1.4 : Dessiner les comptages en barres
# Un graphique se lit plus vite qu un tableau pour voir le desequilibre
# Une figure par dossier : train, validation, test
def charts(count: dict):
    classes = ("NORMAL", "PNEUMONIA")
    fig, axes = plt.subplots(1, 3,  figsize=(12, 5),)
    
    for key, value in count.items():
        i = list(count.keys()).index(key)
        axes[i].bar(classes, value)
        axes[i].set_title(key)
        axes[i].set_xlabel("classes")
        axes[i].set_ylabel(f"{key} count")
    
    plt.tight_layout()
    plt.show()
    
```

```python
# Etape 1.4 : Rassembler les chemins des 3 dossiers
# Un dictionnaire par dossier : cle = classe, valeur = liste des images
# Train, validation et test restent separes depuis le debut

# TRAIN


train_image = {
    "normal": get_images(train_classes_paths, 0),
    "pneumonia": get_images(train_classes_paths, 1)
}

val_image = {
    "normal": get_images(val_classes_paths, 0),
    "pneumonia": get_images(val_classes_paths, 1)
}

test_image = {
    "normal": get_images(test_classes_paths, 0),
    "pneumonia": get_images(test_classes_paths, 1)
}
```

```python
# Etape 1.4 : Compter et afficher la repartition
# Les compteurs viennent des listes deja construites, pas d un nouveau parcours

countes =  {
    "train": [len(train_image["normal"]), len(train_image["pneumonia"])],
    "validation": [len(val_image["normal"]), len(val_image["pneumonia"])],
    "test": [len(test_image["normal"]), len(test_image["pneumonia"])]
}

charts(countes)
```

## 1.5 Identifier les éventuels problèmes

```python
# Etape 1.5 : Tester si une image est lisible
# try/except : une image cassee renvoie False au lieu de faire planter le programme
# img.verify() detecte les fichiers corrompus
def check_image(img_path):
    try:
        with Image.open(img_path) as img:
            img.verify()
            img.convert()
            return True 
    except:
        return False
```

```python
# Etape 1.5 : Appliquer ce test aux 3 ensembles
# Le filtre est ecrit mais commente : le dataset est considere valide en l etat
# Les variables valables pointent sur les memes listes que l entree

train_valid = train_image
# {
#     cls: [img for img in train_image[cls] if check_image(img)]
#     for cls in ["normal", "pneumonia"]
# }

val_valid = val_image
# {
#     cls: [img for img in val_image[cls] if check_image(img)]
#     for cls in ["normal", "pneumonia"]
# }

test_valid =  test_image
# {
#     cls: [img for img in test_image[cls] if check_image(img)]
#     for cls in ["normal", "pneumonia"]
# }

```

## 1.6 Redimensionner les images et les convertir en RGB

```python
# Etape 1.6 : Redimensionner une image et la convertir en RGB
# 180x180 et 3 canaux : c est la forme exacte attendue par le CNN
# Le 255 n est divise qu ici, dans une seule fonction

from tensorflow.keras.utils import img_to_array





def process_image(img):
    image = load_img(
        img,
        target_size=(180, 180),
        color_mode="rgb"
    )
    # if normalize:
    # Division par 255 : c est ICI la normalisation, une seule fois
    image = img_to_array(image) / 255.0
    image = np.asarray(image, dtype=np.float32)
    
    # else:    
    #     image = img_to_array(image)
    return image
    
```

```python
# Etape 1.6 : Appliquer ce traitement a toutes les images
# Chaque liste de chemins devient un tableau NumPy que le modele peut lire
train_processed = {
    cls: np.array([
        process_image(path)
        for path in train_valid[cls]
    ])
    for cls in ["normal", "pneumonia"]
}

val_processed = {
    cls: np.array([
        process_image(path)
        for path in val_valid[cls]
    ])
    for cls in ["normal", "pneumonia"]
}

test_processed = {
    cls: np.array([
        process_image(path)
        for path in test_valid[cls]
    ])
    for cls in ["normal", "pneumonia"]
}
```

## 1.8 Associer les labels aux images

```python
# Etape 1.8 : Assembler les pixels et creer les labels
# NORMAL = 0 et PNEUMONIA = 1 : le label vient du nom du dossier d origine
# On empile NORMAL puis PNEUMONIA pour que pixels et labels restent alignes
# Train
X_train = np.concatenate(
    (
        train_processed["normal"],
        train_processed["pneumonia"]
    ),
    axis=0
)
y_train = np.concatenate(
    (
        np.zeros(len(train_processed["normal"])),
        np.ones(len( train_processed["pneumonia"]))
    ),
    axis= 0
)

# Validation
X_val = np.concatenate(
    (
        val_processed["normal"], 
        val_processed["pneumonia"]
    ), 
    axis=0
)
y_val = np.concatenate(
    (
        np.zeros(len(val_processed["normal"])),
        np.ones(len( val_processed["pneumonia"]))
    ),
    axis= 0
)
# Test
X_test = np.concatenate(
    (
        test_processed["normal"], 
        test_processed["pneumonia"]
    ), 
    axis=0
)
y_test = np.concatenate(
    (
        np.zeros(len(test_processed["normal"])),
        np.ones(len( test_processed["pneumonia"]))
    ),
    axis=0
)
```

```python
# Etape 1.8 : Melanger le train
# Seul le train est melange : sinon le modele verrait tous les NORMAL d abord
# random_state=42 rend le melange identique a chaque execution
from sklearn.utils import shuffle

X_train, y_train = shuffle(X_train, y_train, random_state = 42)
X_train = np.asarray(X_train, dtype=np.float32)
y_train = np.asarray(y_train, dtype=np.float32)
```

## 1.11 Étudier le déséquilibre

```python
# Etape 1.11 : Calculer le poids de chaque classe
# compute_class_weight(balanced) applique total / (2 x nombre de la classe)
# La classe rare pese plus lourd pour eviter de tout predire PNEUMONIA
from sklearn.utils.class_weight import compute_class_weight
 
def equilbre():
    
    classes = np.unique(y_train)


    # Poids utilises plus tard dans model.fit(class_weight=...)
    weights = compute_class_weight(class_weight="balanced", classes=classes, y = y_train)

    class_weights = dict(zip(classes, weights))
    
    print(class_weights)

    return class_weights

class_weights = equilbre()
```

```python
# Etape 1.12 : Construire le modele et le compiler
# 2 convolutions apprennent les motifs, la couche dense tranche ensuite
# Sortie sigmoide : un score entre 0 et 1 = probabilite d etre PNEUMONIA

from tensorflow import keras
from tensorflow.keras import layers

model = keras.Sequential([
    # Entree : hauteur, largeur, 3 canaux RGB
    keras.Input(shape=(180, 180, 3)),

    # Extraction des motifs : chaque Conv2D cherche des formes
    layers.Conv2D(
        filters=32,
        kernel_size=(3, 3),
        activation="relu"
    ),
    # MaxPooling : on divise la taille par 2 et on garde l essentiel
    layers.MaxPooling2D(pool_size=(2, 2)),

    layers.Conv2D(
        filters=64,
        kernel_size=(3, 3),
        activation="relu"
    ),
    layers.MaxPooling2D(pool_size=(2, 2)),

    # Decision : on aplatit la carte en vecteur, puis on vote
    layers.Flatten(),
    # Couche de decision apres l aplatissage
    layers.Dense(128, activation="relu"),
    # Dropout 50 % : evite l overfitting (memoriser le train)
    layers.Dropout(0.5),

    # Sortie binaire : 1 score entre 0 et 1 (0=NORMAL, 1=PNEUMONIA)
    layers.Dense(1, activation="sigmoid")
])

model.compile(
    # adam = l optimiseur qui choisit seul le pas d apprentissage
    optimizer="adam",
    # binary_crossentropy = la loss des deux classes opposees (0 ou 1)
    loss="binary_crossentropy",
    metrics=[
        keras.metrics.BinaryAccuracy(name="accuracy"),
        keras.metrics.Precision(name="precision"),
        keras.metrics.Recall(name="recall"),
        keras.metrics.AUC(name="auc"),
    ]
)

model.summary()
```

```python
# Etape 1.13 : Prevoir l arret anticipe de l entrainement
# patience=3 : on arrete apres 3 epochs sans progresse sur la validation
# restore_best_weights : on garde les poids de la meilleure epoch
from tensorflow.keras.callbacks import EarlyStopping

early_stopping = EarlyStopping(monitor="val_loss", patience=3, restore_best_weights=True)

```

```python
# Etape 1.14 : Entrainer le modele (model.fit)
# class_weight corrige le desequilibre, EarlyStopping evite la memorisation
# La duree est mesuree avec time.time()
import numpy as np
import time 



start= time.time()

history = model.fit(
    X_train,
    y_train,
    epochs=5,
    batch_size=32,
    class_weight = class_weights,
    validation_data=(X_val, y_val),
    callbacks = [early_stopping]
)

end = time.time()

training_duration = start - end

training_duration

```

```python
# Etape 1.15 : Tracer la perte train vs validation
# Un ecart qui grandit entre les deux courbes = surapprentissage
# (overfitting = memoriser le train au lieu d apprendre des regles generales)

import matplotlib.pyplot as plt


plt.figure(figsize=(8, 5))
plt.plot(history.history["loss"], label="Train Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.xlabel("Époque")
plt.ylabel("Loss")
plt.title("Évolution de la fonction de perte")
plt.legend()
plt.grid(True)
plt.show()

# # Courbes d'accuracy
# plt.figure(figsize=(8, 5))
# plt.plot(history.history["accuracy"], label="Train Accuracy")
# plt.plot(history.history["val_accuracy"], label="Validation Accuracy")
# plt.xlabel("Époque")
# plt.ylabel("Accuracy")
# plt.title("Évolution de l'accuracy")
# plt.legend()
# plt.grid(True)
# plt.show()
```

# Partie 2 : Pipeline automatisé (tf.data)

```python
# Etape 2.1 : Importer les bibliotheques du pipeline automatique
# Memes besoins qu en partie 1, plus image_dataset_from_directory
import kagglehub
from pathlib import Path
import numpy as np
from PIL import Image
import tensorflow as tf
from tensorflow.keras.preprocessing import image_dataset_from_directory
import matplotlib.pyplot as plt
from tensorflow.keras.utils import load_img
from sklearn.utils.class_weight import compute_class_weight
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow import keras
from tensorflow.keras import layers
```

```python
# Etape 2.2 : Recharger le meme dataset
# Le dossier est deja en cache : kagglehub ne retelecharge rien
path = kagglehub.dataset_download("paultimothymooney/chest-xray-pneumonia")
path = Path(path)
```

```python
# Etape 2.3 : Definir les dossiers des classes
# Memes chemins qu en partie 1, donnes aux datasets automatiques
train_directory =  path / "chest_xray" / "chest_xray" / "train"

val_directory =  path / "chest_xray" / "chest_xray" / "val"

test_directory =  path / "chest_xray" / "chest_xray" / "test"
```

```python
# Etape 2.4 : Creer les datasets tf.data
# La fonction lit les dossiers, redimensionne et regroupe en lots de 32
# seed=42 pour un melange reproductible, label_mode=binary pour une loss binaire
import tensorflow as tf

train_ds = tf.keras.utils.image_dataset_from_directory(
    train_directory,
    labels="inferred",
    label_mode = 'binary', # should be like loss in model
    color_mode="rgb",
    batch_size=32,
    image_size=(180, 180),
    shuffle=True,
    seed = 42, # like random_state
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    train_directory,
    labels="inferred",
    label_mode = 'binary', # should be like loss in model
    color_mode="rgb",
    batch_size=32,
    image_size=(180, 180),
    shuffle=True,
    seed = 42, # like random_state
)

test_ds = tf.keras.utils.image_dataset_from_directory(
    train_directory,
    labels="inferred",
    label_mode = 'binary', # should be like loss in model
    color_mode="rgb",
    batch_size=32,
    image_size=(180, 180),
    shuffle=True,
    seed = 42, # like random_state
)
```

```python
# Etape 2.6 : Calculer les poids de classe du pipeline
# Meme formule qu en 1.11, mais elle s appuie sur y_train de la partie 1

 
def class_weight():
    
    classes = np.unique(y_train)


    computed_class_weight = compute_class_weight(class_weight="balanced", classes=classes, y = y_train)

    class_weights = dict(zip(classes, computed_class_weight))

    return class_weights

computed_class_weight = class_weight()
```

```python
# Etape 2.7 : Construire le modele avec pretraitement embarque
# Rescaling(1/255) est une couche du modele : les images arrivent en 0-255
# Meme architecture qu en 1.12 sinon



model = keras.Sequential([
    # Entree : hauteur, largeur, 3 canaux RGB
    keras.Input(shape=(180, 180, 3)),
    # Resizing force 180x180, meme si une image arrive avec une autre taille
    layers.Resizing(180, 180),  # new layer, test it
    # Rescaling DANS le modele : les donnees arrivent en 0-255
    layers.Rescaling(1./255), # new layer, test it
     
    # Extraction des motifs : chaque Conv2D cherche des formes
    layers.Conv2D(
        filters=32,
        kernel_size=(3, 3),
        activation="relu"
    ),
    # MaxPooling : on divise la taille par 2 et on garde l essentiel
    layers.MaxPooling2D(pool_size=(2, 2)),

    layers.Conv2D(
        filters=64,
        kernel_size=(3, 3),
        activation="relu"
    ),
    layers.MaxPooling2D(pool_size=(2, 2)),

    # Decision : on aplatit la carte en vecteur, puis on vote
    layers.Flatten(),
    # Couche de decision apres l aplatissage
    layers.Dense(128, activation="relu"),
    # Dropout 50 % : evite l overfitting (memoriser le train)
    layers.Dropout(0.5),

    # Sortie binaire : 1 score entre 0 et 1 (0=NORMAL, 1=PNEUMONIA)
    layers.Dense(1, activation="sigmoid")
])

model.compile(
    # adam = l optimiseur qui choisit seul le pas d apprentissage
    optimizer="adam",
    # binary_crossentropy = la loss des deux classes opposees (0 ou 1)
    loss="binary_crossentropy",
    metrics=[
        keras.metrics.BinaryAccuracy(name="accuracy"),
        keras.metrics.Precision(name="precision"),
        keras.metrics.Recall(name="recall"),
        keras.metrics.AUC(name="auc"),
    ]
)

# model.summary()
```

```python
# Etape 2.8 : Entrainer le modele sur le pipeline automatique
# train_ds fournit deja des lots prets a l emploi, sans boucle manuelle
# EarlyStopping arrete la formation si la validation stagne
import numpy as np
import time 

early_stopping = EarlyStopping(monitor="val_loss", patience=3, restore_best_weights=True)



history = model.fit(
    train_ds,
    epochs=5,
    batch_size=32,
    class_weight = class_weights,
    validation_data=val_ds,
    callbacks = [early_stopping]
)
```
